```python
import os
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import load_model
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# --------------------------------------------------------------
# 🔧 Paths
# --------------------------------------------------------------
BASE_PATH = r"C:\Users\NXTWAVE\Downloads\Hybrid Wind-Turbine Performance Optimizer"
DATA_1 = os.path.join(BASE_PATH, "archive", "GE Turbine Power Curve.csv")
DATA_2 = os.path.join(BASE_PATH, "archive", "TexasTurbine.csv")

MODEL_PATH = os.path.join(BASE_PATH, "hybrid_windwise_model.h5")
SCALER_PATH = os.path.join(BASE_PATH, "hybrid_windwise_scaler.pkl")
CONFIG_PATH = os.path.join(BASE_PATH, "hybrid_windwise_config.yaml")
RESULT_PATH = os.path.join(BASE_PATH, "hybrid_windwise_prediction.json")
VISUALS_PATH = os.path.join(BASE_PATH, "visuals")
os.makedirs(VISUALS_PATH, exist_ok=True)

# --------------------------------------------------------------
# 📥 Load and clean dataset
# --------------------------------------------------------------
print("[INFO] Loading datasets...")
df1 = pd.read_csv(DATA_1, skip_blank_lines=True, on_bad_lines='skip')
df2 = pd.read_csv(DATA_2, skip_blank_lines=True, on_bad_lines='skip')
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()
df = pd.concat([df1, df2], ignore_index=True)
df = df.select_dtypes(include=['float64', 'int64'])
df.dropna(how='all', inplace=True)
print(f"[INFO] Dataset shape: {df.shape}")

# Detect columns automatically
wind_cols = [c for c in df.columns if "wind" in c.lower() and "speed" in c.lower()]
power_cols = [c for c in df.columns if "power" in c.lower() or "kw" in c.lower()]
wind_col = wind_cols[0] if wind_cols else df.columns[0]
power_col = power_cols[0] if power_cols else df.columns[min(1, len(df.columns) - 1)]

df = df[[wind_col, power_col]].copy()
df.columns = ["WindSpeed", "Power"]
df["Cp"] = df["Power"] / (df["WindSpeed"] ** 3 + 1e-5)
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

# --------------------------------------------------------------
# 🧮 Create time-series sequences
# --------------------------------------------------------------
def create_sequences(data, lookback=10):
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:i + lookback])
        y.append(data[i + lookback, 0])  # predict next Power
    return np.array(X), np.array(y)

scaler = joblib.load(SCALER_PATH)
data_scaled = scaler.transform(df)
lookback = 10
X, y = create_sequences(data_scaled, lookback)
split = int(0.8 * len(X))
X_test, y_test = X[split:], y[split:]

# --------------------------------------------------------------
# 🧠 Load model + configs
# --------------------------------------------------------------
model = load_model(MODEL_PATH)
with open(CONFIG_PATH, "r") as f:
    config = yaml.safe_load(f)

# --------------------------------------------------------------
# 🔮 Predict and evaluate
# --------------------------------------------------------------
y_pred = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

# --------------------------------------------------------------
# 🧾 Display results on screen
# --------------------------------------------------------------
print("\n================ WINDWISE RESULTS ================\n")
print(f"Optimizer Type       : {config['optimizer']}")
print(f"Best Parameters      : {config['best_params']}")
print(f"Optimization Score   : {config['best_score']:.6f}")
print(f"\nModel Evaluation:")
print(f"  ➤ RMSE : {rmse:.4f}")
print(f"  ➤ MAE  : {mae:.4f}")
print(f"  ➤ R²   : {r2:.4f}")
print("\n==================================================\n")

# --------------------------------------------------------------
# 📊 1️⃣ Accuracy Graph
# --------------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(y_test[:200], label="Actual", linewidth=2)
plt.plot(y_pred[:200], label="Predicted", linestyle='--', linewidth=2)
plt.title("WindWise Accuracy Graph")
plt.xlabel("Samples")
plt.ylabel("Scaled Power Output")
plt.legend()
plt.show()

# --------------------------------------------------------------
# 📊 2️⃣ Heatmap
# --------------------------------------------------------------
plt.figure(figsize=(6, 5))
sns.heatmap(np.corrcoef(df.T), cmap="coolwarm", annot=False)
plt.title("Feature Correlation Heatmap")
plt.show()

# --------------------------------------------------------------
# 📊 3️⃣ Comparison Graph
# --------------------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(y_test[:300], label="Actual", linewidth=2)
plt.plot(y_pred[:300], label="Predicted", linestyle='dashed', linewidth=2)
plt.title("WindWise Comparison Graph")
plt.xlabel("Time Steps")
plt.ylabel("Scaled Power Output")
plt.legend()
plt.show()

# --------------------------------------------------------------
# 📊 4️⃣ Result Graph (Bar Plot)
# --------------------------------------------------------------
metrics = ["RMSE", "MAE", "R2"]
values = [rmse, mae, r2]
plt.figure(figsize=(7, 5))
sns.barplot(x=metrics, y=values, palette="viridis")
plt.title("Model Result Graph (Performance Metrics)")
plt.xlabel("Metric")
plt.ylabel("Value")
for i, v in enumerate(values):
    plt.text(i, v + 0.01, f"{v:.3f}", ha='center')
plt.show()

# --------------------------------------------------------------
# 📊 5️⃣ Prediction Graph (Full Range)
# --------------------------------------------------------------
plt.figure(figsize=(12, 5))
plt.plot(y_test, label="Actual", linewidth=1)
plt.plot(y_pred, label="Predicted", linewidth=1)
plt.title("WindWise Prediction Graph (Full Series)")
plt.xlabel("Time Steps")
plt.ylabel("Scaled Power Output")
plt.legend()
plt.show()

print("[INFO] ✅ All results displayed successfully!")
```
