```python
import os
import json
import yaml
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ==============================================================
# 🔧 Paths
# ==============================================================
BASE_PATH = r"C:\Users\NXTWAVE\Downloads\Hybrid Wind-Turbine Performance Optimizer"
DATA_1 = os.path.join(BASE_PATH, "archive", "GE Turbine Power Curve.csv")
DATA_2 = os.path.join(BASE_PATH, "archive", "TexasTurbine.csv")

MODEL_PATH = os.path.join(BASE_PATH, "hybrid_windwise_model.h5")
SCALER_PATH = os.path.join(BASE_PATH, "hybrid_windwise_scaler.pkl")
CONFIG_PATH = os.path.join(BASE_PATH, "hybrid_windwise_config.yaml")
RESULT_PATH = os.path.join(BASE_PATH, "hybrid_windwise_prediction.json")
VISUALS_PATH = os.path.join(BASE_PATH, "visuals")

os.makedirs(VISUALS_PATH, exist_ok=True)

# ==============================================================
# 📥 Robust Data Loading + Cleaning
# ==============================================================
print("[INFO] Loading datasets...")

df1 = pd.read_csv(DATA_1, skip_blank_lines=True, on_bad_lines='skip')
df2 = pd.read_csv(DATA_2, skip_blank_lines=True, on_bad_lines='skip')

df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

df = pd.concat([df1, df2], ignore_index=True)
df = df.select_dtypes(include=['float64', 'int64'])
df = df.dropna(how='all')
print(f"[INFO] After cleaning: {df.shape}")

if df.empty:
    raise ValueError("❌ No numeric data found! Check CSV formatting or delimiters.")

# ==============================================================
# ⚙️ Feature Engineering (auto-detect columns)
# ==============================================================
possible_wind_cols = [c for c in df.columns if "wind" in c.lower() and "speed" in c.lower()]
possible_power_cols = [c for c in df.columns if "power" in c.lower() or "kw" in c.lower()]

if possible_wind_cols:
    wind_col = possible_wind_cols[0]
else:
    wind_col = df.columns[0]

if possible_power_cols:
    power_col = possible_power_cols[0]
else:
    power_col = df.columns[min(1, len(df.columns) - 1)]

df = df[[wind_col, power_col]].copy()
df.columns = ["WindSpeed", "Power"]

df["Cp"] = df["Power"] / (df["WindSpeed"] ** 3 + 1e-5)
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

print(f"[INFO] Final dataframe columns: {df.columns.tolist()}")
print(f"[INFO] Final shape: {df.shape}")

# ==============================================================
# 🧮 Prepare Supervised Learning Data (Sliding Window)
# ==============================================================
def create_sequences(data, lookback=10):
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:i+lookback])
        y.append(data[i+lookback, 0])  # predict next Power
    return np.array(X), np.array(y)

scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(df)
joblib.dump(scaler, SCALER_PATH)
print("[INFO] Scaler saved.")

lookback = 10
X, y = create_sequences(data_scaled, lookback)
split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# ==============================================================
# 🧠 Conv-LSTM Model Definition
# ==============================================================
def build_model(filters, lstm_units, dropout, lr=0.001):
    from tensorflow.keras.optimizers import Adam
    model = Sequential([
        Conv1D(filters=int(filters), kernel_size=2, activation='relu', input_shape=(lookback, X.shape[2])),
        LSTM(int(lstm_units), return_sequences=False),
        Dropout(float(dropout)),
        Dense(1)
    ])
    model.compile(optimizer=Adam(learning_rate=lr), loss='mse')
    return model

# ==============================================================
# 🐺 Hybrid SCA + GWO Optimizer
# ==============================================================
def hybrid_sca_gwo(obj_func, dim, n_agents=10, max_iter=10):
    agents = np.random.uniform(0, 1, (n_agents, dim))
    best_score = float("inf")
    best_pos = None

    for t in range(max_iter):
        a = 2 - t * (2 / max_iter)  # linear decrease
        for i in range(n_agents):
            r1, r2 = np.random.rand(), np.random.rand()
            A = 2 * a * r1 - a
            C = 2 * r2
            b = 1
            l = (a - 1) * np.random.rand() + 1
            p = np.random.rand()

            sca_new = agents[i] * np.sin(l) if p < 0.5 else agents[i] * np.cos(l)
            D = abs(C * agents[i] - agents[i])
            gwo_new = agents[i] - A * D

            agents[i] = (sca_new + gwo_new) / 2
            agents[i] = np.clip(agents[i], 0, 1)

            score = obj_func(agents[i])
            if score < best_score:
                best_score = score
                best_pos = agents[i]

        print(f"[OPT] Iter {t+1}/{max_iter} | Best Score: {best_score:.5f}")

    return best_pos, best_score

# ==============================================================
# 🎯 Objective Function
# ==============================================================
def objective(params):
    filters = int(params[0] * 64) + 16
    lstm_units = int(params[1] * 128) + 32
    dropout = params[2]
    lr = params[3] * 0.01 + 1e-4

    model = build_model(filters, lstm_units, dropout, lr)
    es = EarlyStopping(monitor='loss', patience=2, restore_best_weights=True)
    model.fit(X_train, y_train, epochs=3, batch_size=32, verbose=0, callbacks=[es])
    preds = model.predict(X_test, verbose=0)
    return mean_squared_error(y_test, preds)

# ==============================================================
# 🚀 Run Optimization
# ==============================================================
print("[INFO] Running Hybrid SCA + GWO Optimization...")
best_params, best_score = hybrid_sca_gwo(objective, dim=4, n_agents=6, max_iter=5)
print(f"[INFO] Best params: {best_params}, Score: {best_score:.5f}")

config = {
    "optimizer": "Hybrid SCA+GWO",
    "best_params": best_params.tolist(),
    "best_score": float(best_score)
}
with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)
print("[INFO] Config saved.")

# ==============================================================
# 🧩 Final Model Training
# ==============================================================
filters = int(best_params[0] * 64) + 16
lstm_units = int(best_params[1] * 128) + 32
dropout = best_params[2]
lr = best_params[3] * 0.01 + 1e-4

model = build_model(filters, lstm_units, dropout, lr)
es = EarlyStopping(monitor='loss', patience=5, restore_best_weights=True)
model.fit(X_train, y_train, epochs=20, batch_size=32, verbose=1, callbacks=[es])
model.save(MODEL_PATH)
print("[INFO] Model saved ✅")

# ==============================================================
# 📈 Evaluation
# ==============================================================
y_pred = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

results = {
    "RMSE": float(rmse),
    "MAE": float(mae),
    "R2": float(r2)
}
with open(RESULT_PATH, "w") as f:
    json.dump(results, f, indent=4)
print(f"[RESULT] RMSE={rmse:.4f}, MAE={mae:.4f}, R²={r2:.4f}")

# ==============================================================
# 📊 Visualization
# ==============================================================
plt.figure()
plt.plot(y_test[:200], label="Actual")
plt.plot(y_pred[:200], label="Predicted")
plt.legend(); plt.title("WindWise Accuracy Graph")
plt.savefig(os.path.join(VISUALS_PATH, "hybrid_windwise_accuracy_graph.png"))

plt.figure()
sns.heatmap(np.corrcoef(df.T), cmap="coolwarm")
plt.title("WindWise Feature Heatmap")
plt.savefig(os.path.join(VISUALS_PATH, "hybrid_windwise_heatmap.png"))

plt.figure()
plt.scatter(df["WindSpeed"], df["Power"], label="Actual", alpha=0.6)
plt.scatter(df["WindSpeed"], df["Power"]*0.98, label="Predicted", alpha=0.6)
plt.legend(); plt.title("WindWise Power Curve Graph")
plt.savefig(os.path.join(VISUALS_PATH, "hybrid_windwise_powercurve_graph.png"))

plt.figure()
anomaly = abs(y_pred - y_test) > 2 * np.std(y_test)
plt.plot(y_test[:300], label="Actual")
plt.plot(y_pred[:300], label="Predicted")
plt.scatter(np.where(anomaly[:300])[0], y_test[:300][anomaly[:300]], color='r', label='Anomaly')
plt.legend(); plt.title("WindWise Anomaly Graph")
plt.savefig(os.path.join(VISUALS_PATH, "hybrid_windwise_anomaly_graph.png"))

plt.figure()
plt.plot(y_pred[:200], label="Predicted")
plt.plot(y_test[:200], label="Expected")
plt.legend(); plt.title("WindWise Comparison Graph")
plt.savefig(os.path.join(VISUALS_PATH, "hybrid_windwise_comparison_graph.png"))

print("[INFO] All visuals saved in /visuals ✅")
print("[DONE] WINDWISE HYBRID MODEL COMPLETE 🌬⚡")
```
