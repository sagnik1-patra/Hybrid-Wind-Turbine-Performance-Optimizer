```python
import os, json, yaml, joblib, warnings
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
warnings.filterwarnings("ignore")

# --------------------------------------------------------------
# 📁 Paths
# --------------------------------------------------------------
BASE = r"C:\Users\NXTWAVE\Downloads\Hybrid Wind-Turbine Performance Optimizer"
DATA1 = os.path.join(BASE, "archive", "GE Turbine Power Curve.csv")
DATA2 = os.path.join(BASE, "archive", "TexasTurbine.csv")

MODEL = os.path.join(BASE, "ais_csa_windwise_model.h5")
SCALER = os.path.join(BASE, "ais_csa_windwise_scaler.pkl")
CONFIG = os.path.join(BASE, "ais_csa_windwise_config.yaml")
RESULT = os.path.join(BASE, "ais_csa_windwise_result.json")
VISUAL = os.path.join(BASE, "visuals")
os.makedirs(VISUAL, exist_ok=True)

# --------------------------------------------------------------
# 📥 Load + Clean
# --------------------------------------------------------------
print("[INFO] Loading datasets…")
df1 = pd.read_csv(DATA1, on_bad_lines="skip")
df2 = pd.read_csv(DATA2, on_bad_lines="skip")
df1.columns, df2.columns = df1.columns.str.strip(), df2.columns.str.strip()
df = pd.concat([df1, df2], ignore_index=True)
df = df.select_dtypes(include=["float64", "int64"]).dropna(how="all")
if df.empty: raise ValueError("No numeric data found.")
# auto-detect cols
wind = [c for c in df.columns if "wind" in c.lower() and "speed" in c.lower()]
power = [c for c in df.columns if "power" in c.lower() or "kw" in c.lower()]
wind = wind[0] if wind else df.columns[0]
power = power[0] if power else df.columns[min(1,len(df.columns)-1)]
df = df[[wind,power]].copy()
df.columns=["WindSpeed","Power"]
df["Cp"]=df["Power"]/(df["WindSpeed"]**3+1e-5)
df.replace([np.inf,-np.inf],np.nan,inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True,inplace=True)
print(f"[INFO] Data shape: {df.shape}")

# --------------------------------------------------------------
# 🧮 Sequence prep
# --------------------------------------------------------------
def create_seq(data,look=10):
    X,y=[],[]
    for i in range(len(data)-look):
        X.append(data[i:i+look]);y.append(data[i+look,0])
    return np.array(X),np.array(y)

scaler=MinMaxScaler();data_s=scaler.fit_transform(df)
joblib.dump(scaler,SCALER)
look=10
X,y=create_seq(data_s,look)
split=int(0.8*len(X))
Xtr,Xte,ytr,yte=X[:split],X[split:],y[:split],y[split:]

# --------------------------------------------------------------
# 🧠 Model
# --------------------------------------------------------------
def make_model(fil,units,drop,lr):
    from tensorflow.keras.optimizers import Adam
    m=Sequential([
        Conv1D(int(fil),2,activation="relu",input_shape=(look,X.shape[2])),
        LSTM(int(units)),Dropout(drop),Dense(1)])
    m.compile(optimizer=Adam(lr),loss="mse");return m

# --------------------------------------------------------------
# 🧬 AIS + CSA Hybrid Optimizer
# --------------------------------------------------------------
def hybrid_ais_csa(obj,dim=4,agents=6,gen=6):
    pop=np.random.rand(agents,dim)
    fit=np.array([obj(p) for p in pop])
    best_idx=np.argmin(fit)
    best=pop[best_idx].copy();best_f=fit[best_idx]

    for g in range(gen):
        # ---- AIS mutation (clonal selection) ----
        clones=pop+0.1*np.random.randn(*pop.shape)
        clones=np.clip(clones,0,1)
        fcl=np.array([obj(p) for p in clones])
        for i in range(agents):
            if fcl[i]<fit[i]: pop[i],fit[i]=clones[i],fcl[i]
        # ---- CSA update (Levy flight) ----
        beta=1.5; sigma=( (np.math.gamma(1+beta)*np.sin(np.pi*beta/2)) /
                (np.math.gamma((1+beta)/2)*beta*2**((beta-1)/2)) )**(1/beta)
        for i in range(agents):
            step=np.random.randn(dim)*sigma
            new=pop[i]+0.01*step*(pop[i]-best)
            new=np.clip(new,0,1)
            fnew=obj(new)
            if fnew<fit[i]: pop[i],fit[i]=new,fnew
        # ---- update best ----
        bi=np.argmin(fit)
        if fit[bi]<best_f: best,best_f=pop[bi].copy(),fit[bi]
        print(f"[OPT] Gen {g+1}/{gen} | Best Score: {best_f:.5f}")
    return best,best_f

# --------------------------------------------------------------
# 🎯 Objective
# --------------------------------------------------------------
def obj(params):
    f=int(params[0]*64)+16
    u=int(params[1]*128)+32
    d=float(params[2])
    lr=params[3]*0.01+1e-4
    m=make_model(f,u,d,lr)
    es=EarlyStopping(monitor="loss",patience=2,restore_best_weights=True)
    m.fit(Xtr,ytr,epochs=3,batch_size=32,verbose=0,callbacks=[es])
    p=m.predict(Xte,verbose=0)
    return mean_squared_error(yte,p)

# --------------------------------------------------------------
# 🚀 Run Optimization
# --------------------------------------------------------------
print("[INFO] Running AIS+CSA optimization...")
best,best_score=hybrid_ais_csa(obj)
conf={"optimizer":"Hybrid AIS+CSA","best_params":best.tolist(),"best_score":float(best_score)}
with open(CONFIG,"w") as f: yaml.dump(conf,f)

# --------------------------------------------------------------
# 🧩 Train Final Model
# --------------------------------------------------------------
f=int(best[0]*64)+16;u=int(best[1]*128)+32;d=float(best[2]);lr=best[3]*0.01+1e-4
model=make_model(f,u,d,lr)
es=EarlyStopping(monitor="loss",patience=5,restore_best_weights=True)
model.fit(Xtr,ytr,epochs=20,batch_size=32,verbose=1,callbacks=[es])
model.save(MODEL)
print("[INFO] Model saved ✅")

# --------------------------------------------------------------
# 📈 Evaluate
# --------------------------------------------------------------
yp=model.predict(Xte)
rmse=np.sqrt(mean_squared_error(yte,yp))
mae=mean_absolute_error(yte,yp)
r2=r2_score(yte,yp)
res={"RMSE":float(rmse),"MAE":float(mae),"R2":float(r2)}
with open(RESULT,"w") as f: json.dump(res,f,indent=4)
print(f"[RESULT] RMSE={rmse:.4f}, MAE={mae:.4f}, R²={r2:.4f}")

# --------------------------------------------------------------
# 📊 Visuals
# --------------------------------------------------------------
# 1 Accuracy
plt.figure();plt.plot(yte[:200],label="Actual");plt.plot(yp[:200],label="Pred")
plt.legend();plt.title("AIS+CSA WindWise Accuracy");plt.savefig(os.path.join(VISUAL,"ais_csa_windwise_accuracy_graph.png"))
# 2 Heatmap
plt.figure();sns.heatmap(np.corrcoef(df.T),cmap="coolwarm")
plt.title("AIS+CSA Feature Heatmap");plt.savefig(os.path.join(VISUAL,"ais_csa_windwise_heatmap.png"))
# 3 Comparison
plt.figure();plt.plot(yp[:300],label="Pred");plt.plot(yte[:300],label="Actual")
plt.legend();plt.title("AIS+CSA WindWise Comparison");plt.savefig(os.path.join(VISUAL,"ais_csa_windwise_comparison_graph.png"))
# 4 Result
plt.figure();sns.barplot(x=["RMSE","MAE","R2"],y=[rmse,mae,r2],palette="viridis")
plt.title("AIS+CSA WindWise Result Graph");plt.savefig(os.path.join(VISUAL,"ais_csa_windwise_result_graph.png"))
# 5 Prediction
plt.figure();plt.plot(yte,label="Actual");plt.plot(yp,label="Pred");plt.legend()
plt.title("AIS+CSA WindWise Prediction Graph");plt.savefig(os.path.join(VISUAL,"ais_csa_windwise_prediction_graph.png"))
print("[INFO] ✅ Visuals saved in /visuals")
print("[DONE] WINDWISE HYBRID AIS+CSA COMPLETE 🌬⚡")
```
