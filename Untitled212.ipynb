```python
import os, json, yaml, joblib, warnings
import numpy as np, pandas as pd
from tensorflow.keras.models import load_model
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
warnings.filterwarnings("ignore")

# --------------------------------------------------------------
# 📁 Paths
# --------------------------------------------------------------
BASE = r"C:\Users\NXTWAVE\Downloads\Hybrid Wind-Turbine Performance Optimizer"
DATA_PATH = os.path.join(BASE, "archive", "GE Turbine Power Curve.csv")

MODEL_PATH = os.path.join(BASE, "gwoa_woa_windwise_model.h5")
SCALER_PATH = os.path.join(BASE, "gwoa_woa_windwise_scaler.pkl")
CONFIG_PATH = os.path.join(BASE, "gwoa_woa_windwise_config.yaml")

PRED_JSON = os.path.join(BASE, "gwoa_woa_windwise_prediction.json")
RESULT_CSV = os.path.join(BASE, "gwoa_woa_windwise_result.csv")

# --------------------------------------------------------------
# 📥 Load Data
# --------------------------------------------------------------
print("[INFO] Loading dataset for prediction...")
df = pd.read_csv(DATA_PATH, on_bad_lines='skip')
df.columns = df.columns.str.strip()
df = df.select_dtypes(include=["float64", "int64"]).dropna(how="all")
if df.empty:
    raise ValueError("❌ No numeric columns found in dataset!")

# Auto-detect wind/power columns
wind = [c for c in df.columns if "wind" in c.lower() and "speed" in c.lower()]
power = [c for c in df.columns if "power" in c.lower() or "kw" in c.lower()]
wind = wind[0] if wind else df.columns[0]
power = power[0] if power else df.columns[min(1, len(df.columns) - 1)]
df = df[[wind, power]].copy()
df.columns = ["WindSpeed", "Power"]

df["Cp"] = df["Power"] / (df["WindSpeed"] ** 3 + 1e-5)
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)
print(f"[INFO] Cleaned dataset shape: {df.shape}")

# --------------------------------------------------------------
# 🧮 Sequence Preparation
# --------------------------------------------------------------
def create_seq(data, look=10):
    X, y = [], []
    for i in range(len(data) - look):
        X.append(data[i:i + look])
        y.append(data[i + look, 0])  # predict next Power
    return np.array(X), np.array(y)

scaler = joblib.load(SCALER_PATH)
data_scaled = scaler.transform(df)
look = 10
X, y = create_seq(data_scaled, look)

# --------------------------------------------------------------
# 🧠 Load Model + Config
# --------------------------------------------------------------
model = load_model(MODEL_PATH)
with open(CONFIG_PATH, "r") as f:
    config = yaml.safe_load(f)
print("[INFO] Loaded trained model and optimizer config ✅")

# --------------------------------------------------------------
# 🔮 Make Predictions
# --------------------------------------------------------------
y_pred = model.predict(X)
y_pred = y_pred.flatten()
y_true = y.flatten()

# --------------------------------------------------------------
# 📈 Evaluate
# --------------------------------------------------------------
rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
mae = float(mean_absolute_error(y_true, y_pred))
r2 = float(r2_score(y_true, y_pred))
metrics = {"RMSE": rmse, "MAE": mae, "R2": r2}

# --------------------------------------------------------------
# 💾 Save Prediction JSON
# --------------------------------------------------------------
pred_df = pd.DataFrame({
    "Actual_Power": y_true,
    "Predicted_Power": y_pred
})
pred_json = {
    "optimizer": "Hybrid GWOA+WOA",
    "config": config,
    "metrics": metrics,
    "predictions": pred_df.head(50).to_dict(orient="records")  # store first 50 to JSON
}
with open(PRED_JSON, "w") as f:
    json.dump(pred_json, f, indent=4)
print(f"[INFO] Saved predictions → {PRED_JSON}")

# --------------------------------------------------------------
# 💾 Save Result CSV
# --------------------------------------------------------------
result_df = pred_df.copy()
result_df["Error"] = abs(result_df["Actual_Power"] - result_df["Predicted_Power"])
summary = pd.DataFrame([metrics])
summary["Optimizer"] = "Hybrid GWOA+WOA"
summary["Best_Params"] = [config["best_params"]]
summary["Best_Score"] = [config["best_score"]]
final_csv = pd.concat([summary, result_df.head(500)], axis=1)
final_csv.to_csv(RESULT_CSV, index=False)
print(f"[INFO] Saved result CSV → {RESULT_CSV}")

# --------------------------------------------------------------
# ✅ Print Summary
# --------------------------------------------------------------
print("\n================ WINDWISE GWOA+WOA PREDICTION ================")
print(f"Optimizer Type : {config['optimizer']}")
print(f"Best Params    : {config['best_params']}")
print(f"Best Score     : {config['best_score']:.6f}")
print("\nModel Evaluation:")
print(f"  ➤ RMSE : {rmse:.4f}")
print(f"  ➤ MAE  : {mae:.4f}")
print(f"  ➤ R²   : {r2:.4f}")
print("==============================================================\n")
print("[DONE] ✅ Prediction JSON and Result CSV generated successfully 🌬⚡")
```
