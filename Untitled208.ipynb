```python
import os
import json
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import load_model
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# ==============================================================
# 🔧 Paths
# ==============================================================
BASE_PATH = r"C:\Users\NXTWAVE\Downloads\Hybrid Wind-Turbine Performance Optimizer"
DATA_1 = os.path.join(BASE_PATH, "archive", "GE Turbine Power Curve.csv")
DATA_2 = os.path.join(BASE_PATH, "archive", "TexasTurbine.csv")

MODEL_PATH = os.path.join(BASE_PATH, "hybrid_windwise_model.h5")
SCALER_PATH = os.path.join(BASE_PATH, "hybrid_windwise_scaler.pkl")
CONFIG_PATH = os.path.join(BASE_PATH, "hybrid_windwise_config.yaml")
RESULT_PATH = os.path.join(BASE_PATH, "hybrid_windwise_prediction.json")
VISUALS_PATH = os.path.join(BASE_PATH, "visuals")

os.makedirs(VISUALS_PATH, exist_ok=True)

# ==============================================================
# 📥 Load Data
# ==============================================================
print("[INFO] Loading datasets...")
df1 = pd.read_csv(DATA_1, skip_blank_lines=True, on_bad_lines='skip')
df2 = pd.read_csv(DATA_2, skip_blank_lines=True, on_bad_lines='skip')
df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

df = pd.concat([df1, df2], ignore_index=True)
df = df.select_dtypes(include=['float64', 'int64'])
df.dropna(how='all', inplace=True)
print(f"[INFO] Dataset shape after cleaning: {df.shape}")

# Detect likely wind/power columns
possible_wind_cols = [c for c in df.columns if "wind" in c.lower() and "speed" in c.lower()]
possible_power_cols = [c for c in df.columns if "power" in c.lower() or "kw" in c.lower()]

if possible_wind_cols:
    wind_col = possible_wind_cols[0]
else:
    wind_col = df.columns[0]

if possible_power_cols:
    power_col = possible_power_cols[0]
else:
    power_col = df.columns[min(1, len(df.columns)-1)]

df = df[[wind_col, power_col]].copy()
df.columns = ["WindSpeed", "Power"]
df["Cp"] = df["Power"] / (df["WindSpeed"] ** 3 + 1e-5)
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)

# ==============================================================
# 🧮 Prepare Supervised Learning Data
# ==============================================================
def create_sequences(data, lookback=10):
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:i+lookback])
        y.append(data[i+lookback, 0])  # predict next power
    return np.array(X), np.array(y)

scaler = joblib.load(SCALER_PATH)
data_scaled = scaler.transform(df)
lookback = 10
X, y = create_sequences(data_scaled, lookback)
split = int(0.8 * len(X))
X_test, y_test = X[split:], y[split:]

# ==============================================================
# 🧠 Load Model + Config + Results
# ==============================================================
model = load_model(MODEL_PATH)
with open(CONFIG_PATH, "r") as f:
    config = yaml.safe_load(f)
with open(RESULT_PATH, "r") as f:
    results = json.load(f)

print(f"[INFO] Loaded model and configuration.")
print(f"[INFO] RMSE: {results['RMSE']:.4f}, MAE: {results['MAE']:.4f}, R²: {results['R2']:.4f}")

# ==============================================================
# 🔮 Predictions
# ==============================================================
y_pred = model.predict(X_test)
y_pred_inv = y_pred
y_test_inv = y_test

# ==============================================================
# 📊 1️⃣ Accuracy Graph
# ==============================================================
plt.figure(figsize=(10, 5))
plt.plot(y_test_inv[:200], label='Actual', linewidth=2)
plt.plot(y_pred_inv[:200], label='Predicted', linestyle='dashed')
plt.title("WindWise Accuracy Graph")
plt.xlabel("Sample")
plt.ylabel("Scaled Power Output")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_PATH, "windwise_accuracy_graph.png"))
plt.close()

# ==============================================================
# 📊 2️⃣ Heatmap
# ==============================================================
plt.figure(figsize=(6, 5))
sns.heatmap(np.corrcoef(df.T), cmap="coolwarm", annot=False)
plt.title("WindWise Feature Correlation Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_PATH, "windwise_heatmap.png"))
plt.close()

# ==============================================================
# 📊 3️⃣ Comparison Graph
# ==============================================================
plt.figure(figsize=(10, 5))
plt.plot(y_pred_inv[:300], label="Predicted", linewidth=2)
plt.plot(y_test_inv[:300], label="Actual", linewidth=2, linestyle='dashed')
plt.title("WindWise Comparison Graph")
plt.xlabel("Time Step")
plt.ylabel("Scaled Power")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_PATH, "windwise_comparison_graph.png"))
plt.close()

# ==============================================================
# 📊 4️⃣ Result Graph (Performance Metrics)
# ==============================================================
metrics = ["RMSE", "MAE", "R2"]
values = [results["RMSE"], results["MAE"], results["R2"]]
plt.figure(figsize=(7, 5))
sns.barplot(x=metrics, y=values, palette="viridis")
plt.title("WindWise Result Graph (Model Metrics)")
plt.xlabel("Metric")
plt.ylabel("Score")
for i, v in enumerate(values):
    plt.text(i, v + 0.01, f"{v:.3f}", ha='center')
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_PATH, "windwise_result_graph.png"))
plt.close()

# ==============================================================
# 📊 5️⃣ Prediction Graph (Full Actual vs Predicted)
# ==============================================================
plt.figure(figsize=(12, 5))
plt.plot(y_test_inv, label="Actual", linewidth=1)
plt.plot(y_pred_inv, label="Predicted", linewidth=1)
plt.title("WindWise Prediction Graph (Full Series)")
plt.xlabel("Time Steps")
plt.ylabel("Scaled Power Output")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(VISUALS_PATH, "windwise_prediction_graph.png"))
plt.close()

# ==============================================================
# ✅ Save Summary
# ==============================================================
summary = {
    "AccuracyGraph": "windwise_accuracy_graph.png",
    "Heatmap": "windwise_heatmap.png",
    "ComparisonGraph": "windwise_comparison_graph.png",
    "ResultGraph": "windwise_result_graph.png",
    "PredictionGraph": "windwise_prediction_graph.png"
}
with open(os.path.join(VISUALS_PATH, "windwise_graph_summary.json"), "w") as f:
    json.dump(summary, f, indent=4)

print("[INFO] ✅ All graphs generated successfully!")
print(f"[PATH] Saved in: {VISUALS_PATH}")
```
