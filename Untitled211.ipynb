```python
import os, json, yaml, joblib, warnings
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# --------------------------------------------------------------
# 📁 Paths
# --------------------------------------------------------------
BASE = r"C:\Users\NXTWAVE\Downloads\Hybrid Wind-Turbine Performance Optimizer"
DATA1 = os.path.join(BASE, "archive", "GE Turbine Power Curve.csv")
DATA2 = os.path.join(BASE, "archive", "TexasTurbine.csv")

MODEL = os.path.join(BASE, "gwoa_woa_windwise_model.h5")
SCALER = os.path.join(BASE, "gwoa_woa_windwise_scaler.pkl")
CONFIG = os.path.join(BASE, "gwoa_woa_windwise_config.yaml")
RESULT = os.path.join(BASE, "gwoa_woa_windwise_result.json")
VISUAL = os.path.join(BASE, "visuals")
os.makedirs(VISUAL, exist_ok=True)

# --------------------------------------------------------------
# 📥 Load + Clean
# --------------------------------------------------------------
print("[INFO] Loading datasets...")
df1 = pd.read_csv(DATA1, on_bad_lines='skip')
df2 = pd.read_csv(DATA2, on_bad_lines='skip')
df1.columns, df2.columns = df1.columns.str.strip(), df2.columns.str.strip()
df = pd.concat([df1, df2], ignore_index=True)
df = df.select_dtypes(include=["float64", "int64"]).dropna(how="all")
if df.empty:
    raise ValueError("No numeric data found!")

wind = [c for c in df.columns if "wind" in c.lower() and "speed" in c.lower()]
power = [c for c in df.columns if "power" in c.lower() or "kw" in c.lower()]
wind = wind[0] if wind else df.columns[0]
power = power[0] if power else df.columns[min(1, len(df.columns) - 1)]
df = df[[wind, power]].copy()
df.columns = ["WindSpeed", "Power"]
df["Cp"] = df["Power"] / (df["WindSpeed"] ** 3 + 1e-5)
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.dropna(inplace=True)
df.reset_index(drop=True, inplace=True)
print(f"[INFO] Cleaned data shape: {df.shape}")

# --------------------------------------------------------------
# 🧮 Sequence Prep
# --------------------------------------------------------------
def create_seq(data, look=10):
    X, y = [], []
    for i in range(len(data) - look):
        X.append(data[i:i + look])
        y.append(data[i + look, 0])  # predict next power
    return np.array(X), np.array(y)

scaler = MinMaxScaler()
data_s = scaler.fit_transform(df)
joblib.dump(scaler, SCALER)
look = 10
X, y = create_seq(data_s, look)
split = int(0.8 * len(X))
Xtr, Xte, ytr, yte = X[:split], X[split:], y[:split], y[split:]

# --------------------------------------------------------------
# 🧠 Conv-LSTM Model
# --------------------------------------------------------------
def make_model(fil, units, drop, lr):
    from tensorflow.keras.optimizers import Adam
    m = Sequential([
        Conv1D(int(fil), 2, activation="relu", input_shape=(look, X.shape[2])),
        LSTM(int(units)),
        Dropout(drop),
        Dense(1)
    ])
    m.compile(optimizer=Adam(lr), loss="mse")
    return m

# --------------------------------------------------------------
# 🐺 + 🐋 Hybrid GWOA + WOA Optimizer
# --------------------------------------------------------------
def hybrid_gwoa_woa(obj_func, dim=4, n_agents=6, max_iter=6):
    # Initialize wolves/whales
    pop = np.random.rand(n_agents, dim)
    fit = np.array([obj_func(p) for p in pop])
    alpha, beta, delta = pop[np.argsort(fit)[:3]]
    alpha_score = np.min(fit)

    for t in range(max_iter):
        a = 2 - 2 * (t / max_iter)
        for i in range(n_agents):
            r1, r2 = np.random.rand(), np.random.rand()
            A1, C1 = 2 * a * r1 - a, 2 * r2
            D_alpha = abs(C1 * alpha - pop[i])
            X1 = alpha - A1 * D_alpha

            # WOA update (spiral + encircling)
            p = np.random.rand()
            b = 1
            l = np.random.uniform(-1, 1)
            if p < 0.5:
                D = abs(C1 * alpha - pop[i])
                pop[i] = alpha - A1 * D
            else:
                distance2Leader = abs(alpha - pop[i])
                pop[i] = distance2Leader * np.exp(b * l) * np.cos(2 * np.pi * l) + alpha

            pop[i] = np.clip((pop[i] + X1) / 2, 0, 1)
            score = obj_func(pop[i])
            if score < fit[i]:
                fit[i] = score
                if score < alpha_score:
                    alpha, alpha_score = pop[i].copy(), score

        print(f"[OPT] Iter {t+1}/{max_iter} | Best Score: {alpha_score:.5f}")
    return alpha, alpha_score

# --------------------------------------------------------------
# 🎯 Objective Function
# --------------------------------------------------------------
def objective(params):
    f = int(params[0] * 64) + 16
    u = int(params[1] * 128) + 32
    d = float(params[2])
    lr = params[3] * 0.01 + 1e-4
    m = make_model(f, u, d, lr)
    es = EarlyStopping(monitor="loss", patience=2, restore_best_weights=True)
    m.fit(Xtr, ytr, epochs=3, batch_size=32, verbose=0, callbacks=[es])
    p = m.predict(Xte, verbose=0)
    return mean_squared_error(yte, p)

# --------------------------------------------------------------
# 🚀 Run Hybrid Optimization
# --------------------------------------------------------------
print("[INFO] Running Hybrid GWOA + WOA Optimization...")
best, best_score = hybrid_gwoa_woa(objective)
config = {"optimizer": "Hybrid GWOA+WOA", "best_params": best.tolist(), "best_score": float(best_score)}
with open(CONFIG, "w") as f:
    yaml.dump(config, f)
print("[INFO] Config saved ✅")

# --------------------------------------------------------------
# 🧩 Final Model Training
# --------------------------------------------------------------
f = int(best[0] * 64) + 16
u = int(best[1] * 128) + 32
d = float(best[2])
lr = best[3] * 0.01 + 1e-4

model = make_model(f, u, d, lr)
es = EarlyStopping(monitor="loss", patience=5, restore_best_weights=True)
model.fit(Xtr, ytr, epochs=20, batch_size=32, verbose=1, callbacks=[es])
model.save(MODEL)
print("[INFO] Model saved ✅")

# --------------------------------------------------------------
# 📈 Evaluation
# --------------------------------------------------------------
yp = model.predict(Xte)
rmse = np.sqrt(mean_squared_error(yte, yp))
mae = mean_absolute_error(yte, yp)
r2 = r2_score(yte, yp)
res = {"RMSE": float(rmse), "MAE": float(mae), "R2": float(r2)}
with open(RESULT, "w") as f:
    json.dump(res, f, indent=4)
print(f"[RESULT] RMSE={rmse:.4f}, MAE={mae:.4f}, R²={r2:.4f}")

# --------------------------------------------------------------
# 📊 Visualization
# --------------------------------------------------------------
# 1️⃣ Accuracy Graph
plt.figure()
plt.plot(yte[:200], label="Actual")
plt.plot(yp[:200], label="Predicted")
plt.legend()
plt.title("GWOA+WOA WindWise Accuracy")
plt.savefig(os.path.join(VISUAL, "gwoa_woa_windwise_accuracy_graph.png"))

# 2️⃣ Heatmap
plt.figure()
sns.heatmap(np.corrcoef(df.T), cmap="coolwarm")
plt.title("GWOA+WOA Feature Heatmap")
plt.savefig(os.path.join(VISUAL, "gwoa_woa_windwise_heatmap.png"))

# 3️⃣ Comparison Graph
plt.figure()
plt.plot(yp[:300], label="Predicted")
plt.plot(yte[:300], label="Actual")
plt.legend()
plt.title("GWOA+WOA WindWise Comparison Graph")
plt.savefig(os.path.join(VISUAL, "gwoa_woa_windwise_comparison_graph.png"))

# 4️⃣ Result Graph
plt.figure()
sns.barplot(x=["RMSE", "MAE", "R2"], y=[rmse, mae, r2], palette="mako")
plt.title("GWOA+WOA WindWise Result Graph")
plt.savefig(os.path.join(VISUAL, "gwoa_woa_windwise_result_graph.png"))

# 5️⃣ Prediction Graph
plt.figure()
plt.plot(yte, label="Actual", linewidth=1)
plt.plot(yp, label="Predicted", linewidth=1)
plt.legend()
plt.title("GWOA+WOA WindWise Prediction Graph")
plt.savefig(os.path.join(VISUAL, "gwoa_woa_windwise_prediction_graph.png"))

print("[INFO] ✅ All visuals saved.")
print("[DONE] WINDWISE HYBRID GWOA+WOA COMPLETE 🌬⚡")
```
